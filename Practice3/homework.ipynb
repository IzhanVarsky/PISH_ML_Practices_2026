# Домашнее задание №3: Titanic — сравнение моделей и соревнование

Данные в этой папке:
- `titanic_train.csv` — обучающая выборка **с** меткой `Survived` (как `train.csv` на Kaggle)
- `test.csv` — **тестовая выборка Kaggle без меток**; по ней нужно сделать предсказания и отправить submission

Соревнование: https://www.kaggle.com/competitions/titanic/overview

## Задача

По признакам пассажира предсказать `Survived` (0/1) для объектов из `test.csv`.

## Формат сдачи и лидерборд

1. Ноутбук с воспроизводимым кодом + файл `submission.csv`.
2. **Обязательно** загружаете submission на Kaggle.
3. На защите показываете **Kaggle public score** (скрин / страница submission).
4. Преподаватель заносит **этот Kaggle-скор** в Google-таблицу группы (лидерборд курса).
5. На сдачу — один финальный сабмит / один скор в таблице.

Официальный тест для оценки — **тот, что выдал Kaggle** (`test.csv`), а не ваш собственный hold-out из `train`.

---

## Что нужно сделать

### 1. Загрузка и EDA
- Загрузите `titanic_train.csv` и посмотрите на `test.csv`.
- Типы столбцов, распределения, странности.
- Пропуски: где NaN, как заполняете (и на train, и потом тем же способом на test).

### 2. Подготовка признаков
- Отбор признаков (что выкинуть и почему — коротко).
- Кодирование категориальных.
- Масштабирование числовых — если нужно модели (logreg, SVM, kNN…).
- Важно: пайплайн должен одинаково применяться к `test.csv` (без утечки меток, без fit на test).

### 3. Визуализация
- PCA (или другой метод) до 2D на train, точки с цветом Survived.

### 4. Сравнение моделей (на train)
Официальных меток для `test.csv` у вас нет — поэтому **сравнивать модели** нужно на `titanic_train.csv`:
через кросс-валидацию и/или аккуратный validation-split **только для выбора модели**.

Обучите и сравните **не меньше 5 моделей из разных семейств**.
Нельзя сдать работу с одной только моделью.

Минимальный набор семейств (по одной из каждой группы, можно больше):

1. Линейная модель: `LogisticRegression` (или аналог)
2. Дерево: `DecisionTreeClassifier`
3. Бэггинг: `RandomForestClassifier`
4. Бустинг: хотя бы один из `GradientBoostingClassifier` / `XGBClassifier` / `LGBMClassifier` / `CatBoostClassifier`
5. Ещё что-то другое: например `KNeighborsClassifier` или `SVC`

Для сравнения удобно: accuracy / f1 по CV (или по validation), сводная таблица.
По желанию: confusion matrix на validation.

### 5. Итоговая таблица и вывод
- Сводная таблица: модель → метрики на CV/validation.
- Какую модель берёте в финальный Kaggle-сабмит и почему.
- Где видно переобучение (если смотрели train vs validation).

### 6. Kaggle-сабмит (обязательно)
1. Обучите **финальную** модель на всём `titanic_train.csv`.
2. Предскажите метки для `test.csv`.
3. Сохраните `submission.csv` в формате Kaggle: столбцы `PassengerId`, `Survived`.
4. Загрузите на страницу соревнования Titanic.
5. На защите покажите скрин/страницу submission.

В конце ноутбука:

**Final model: `<название>`**
**Kaggle public score: `0.xxxx`** ← это число идёт в Google-таблицу
